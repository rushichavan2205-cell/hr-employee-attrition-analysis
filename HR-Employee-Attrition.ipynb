# Univariante Analysis  

```python
df.head()
```

```python
sns.countplot(x=df.Gender)
```

```python
categorical=[]

for i in df:
    if len(df[i].unique())<=15:
        categorical.append(i)
```

```python
categorical
```

```python
df_cat=df[categorical]
df_cat.head()
```

```python
continuous=[]

for i in df:
    if len(df[i].unique())>=15:
        continuous.append(i)
```

```python
continuous
```

```python
df_cont=df[continuous]
df_cont.head()
```

```python
df_cat = df.select_dtypes(include='object')
```

```python
plotnumber = 1
plt.figure(figsize=(20,20))
for i in df_cat:
    plt.subplot(4,4,plotnumber)  
    sns.countplot(x=df_cat[i])
    plt.xticks(rotation=45)
    plotnumber += 1
plt.tight_layout()
plt.show()
```

```python
df_cont = df.select_dtypes(exclude='object')
```

```python
plotnumber = 1
plt.figure(figsize=(20,25))
for i in df_cont:
    plt.subplot(7,4,plotnumber)
    sns.histplot(df_cont[i], kde=True)
    plt.xticks(rotation=45)
    plotnumber += 1
plt.tight_layout()
plt.show()
```

# Bivariate Analysis

```python
df.head()
```

## Categorical vs categorical

```python
plotnumber = 1
plt.figure(figsize=(20,25))
for i in df_cat:
    if i != 'Attrition':
        plt.subplot(6,4,plotnumber)
        sns.countplot(x=df_cat[i], hue=df['Attrition'])
        plt.xticks(rotation=45)
        plotnumber += 1
plt.tight_layout()
plt.show()
```

## Numerical vs categorical

```python
plotnumber = 1
plt.figure(figsize=(20,25))
for i in df_cont:
    plt.subplot(7,4,plotnumber)
    sns.boxplot(x=df['Attrition'], y=df_cont[i])
    plotnumber += 1
plt.tight_layout()
plt.show()
```
